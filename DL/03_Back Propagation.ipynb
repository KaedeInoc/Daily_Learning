```python
import torch
x_data=[1.0,2.0,3.0]
y_data=[2.0,4.0,6.0]

w=torch.Tensor([1.0])
w.requires_grad =True
a=0.01

def forward(x):
    return x*w

def loss(x,y):
    y_pred=forward(x)
    return (y_pred-y)**2

print('predict(before training):',4,forward(4).item)

for epoch in range(100):
    for x,y in zip(x_data,y_data):
        l=loss(x,y) #calculate loss
        l.backward() #backward and calculate grad
        print('\tgrad:',x,y,w.grad.item())
        with torch.no_grad():
            w-=a*w.grad #use grad to update
        w.grad.data.zero_()
    
    print('progress:',epoch,l.item())

print('predict(after training):',4,forward(4).item)

```

```python
import torch
x_data=[1.0,2.0,3.0]
y_data=[2.0,4.0,6.0]
a=0.001
b=0.05
c=0.05

w=torch.Tensor([1.0,1.0,2.0])
w.requires_grad=True

def forward(x):
    return w[0]*x**2+w[1]*x+w[2]

def loss(x,y):
    y_pred=forward(x)
    return (y_pred-y)**2

print('predict(before training):',4,forward(4).item())

for epoch in range(10000):
    for x,y in zip(x_data,y_data):
        lr=loss(x,y)
        lr.backward()
        with torch.no_grad():
            w[0]-=a*w.grad[0]
            w[1]-=b*w.grad[1]
            w[2]-=c*w.grad[2]
        w.grad.data.zero_()
        
    print('progress:',epoch,lr.item())
    
print('predict(after training):',4,forward(4).item())
        
```
