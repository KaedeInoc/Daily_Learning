```python
#04_Linear Regression with PyTorch

import torch

#p1:prepare dataset
x_data=torch.Tensor([[1.0],[2.0],[3.0]])
y_data=torch.Tensor([[2.0],[4.0],[6.0]])

#p2:design a linear model

class LinearModel(torch.nn.Module):
    def __init__(self):
        super().__init__()
        self.linear=torch.nn.Linear(1,1)#feature of each input and output(feature->column)
    
    def forward(self,x):#override,前馈计算y_hat
        y_pred=self.linear(x)
        return y_pred

model=LinearModel()#实例化，callable

#p3:construct loss and optimizer

criterion=torch.nn.MSELoss(size_average=False)#mini-batch中的数据量不足时的状况
optimizer=torch.optim.SGD(model.parameters(),lr=0.02) 

#p4:training circle
#forward(for y_pred and loss)->backward(for gradient)->update weight
for epoch in range(100):
    y_pred=model(x_data)
    loss=criterion(y_pred,y_data)
    print(epoch,loss.data)
    
    optimizer.zero_grad()
    loss.backward()
    
    optimizer.step()

#Test    
print('w=',model.linear.weight.item())
print('b=',model.linear.bias.item())

x_test=torch.Tensor([4.0])
y_test=model(x_test)

print('y_test= ',y_test.data,'after training')
```
