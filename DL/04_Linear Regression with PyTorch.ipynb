```python
import torch 
#p1--prepare dataset
x_data=torch.Tensor([[1.0],[2.0],[3.0]])
y_data=torch.Tensor([[2.0],[4.0],[6.0]])

#p2--design a linear model
class LinearModel(torch.nn.Module):
    def __init__(self): #construct function for initial
        super().__init__()
        self.linear=torch.nn.Linear(1,1)#size of each input sample  and each output sample
    def forward(self,x):#just like override.
        y_pred=self.linear(x)
        return y_pred
    
model=LinearModel()#实例化

```

#### class torch.nn.linear(in_features,out_features,bias=True)

*行为样本（维度），列为特征，所以括号内的参数其实是输入输出的列数*

*继承Module类，实例化，即为可调用的对象(__call__(),callable)*

```python

#p3--construct loss and optimizer

criterion=torch.nn.MSELoss(size_average=False)
optimizer=torch.optim.SGD(model.parameters(),lr=0.02)
```

#### class torch.nn.MSELoss(size_average=True,reduce=True)

*1.size_average=False,在某个mini-batch中的样本不足指定数时，可以忽略，因此不用做均值运算（即，size_average就是在最后加一个$\frac{1}{N}$的运算）*

*2.reduce:降维，最后loss的结果要求零维张量（因为在loss的运算中，输出的结果是一个n维的张量，但要求的结果是零维，用到降维）*

*涉及创建计算图的步骤才会需要继承module*

#### class torch.optim.SGD(params,lr=<object,object>,momentum=0,dampening=0,weight_decay=0)

*1.params负责告诉优化器哪些参数需要进行随机梯度下降(拿到模型所有参数)*

*2.lr:学习率（learning rate）,即前文$w=w-a\frac{\partial loss}{\partial w}$中的a*

```python
#p4--training cycle
for epoch in range(500):
    y_pred=model(x_data)
    loss=criterion(y_pred,y_data)
    print(epoch,loss)
    
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()#update
```

*1.forward: 计算$\hat{y}$，计算loss*

*2.backward:计算梯度（先清零grad,框架要求）*

*3.update*

```python
#Output weight and bias
print('w= ',model.linear.weight.item())
print('b= ',model.linear.bias.item())

#Test Model
x_test=torch.Tensor([4.0])
y_test=model(x_test)
print('y= ',y_test.data)
```
