**数据集准备**
setname=torchvision.datasets.MNIST(root='...',train=...,download=...)

MINIST:经典手写数字数据集，可根据目标替换为其他数据集，但只限于torchvision中打包的（大部分都有）

root:数据集存储的位置（期望下载或者事先下载的位置）

train:是否为训练集（True:训练集;False:测试集）

download:是否通过torchvision联网下载数据集(True:是;False:事先已经下载过了，无需下载)
