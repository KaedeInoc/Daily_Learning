## 3. 数据清洗

### 3.1 知识准备：缺失值处理

缺失值是数据分析中最常见的问题之一。Pandas 提供了三个核心方法：

````python
# 创建一个含缺失值的小 DataFrame
import numpy as np
df_miss = pd.DataFrame({
    'A': [1, 2, np.nan, 4],
    'B': [np.nan, 'x', 'y', 'z'],
    'C': [10, 20, 30, 40]
})
df_miss
````

````python
# 查看每列缺失值数量
df_miss.isnull().sum()
````

````python
# 填充数值列的缺失值（用均值）
df_miss['A'].fillna(df_miss['A'].mean(), inplace=True)
df_miss
````

````python
# 删除某一列
df_miss.drop('B', axis=1, inplace=True)
df_miss
````

> ✅ 缺失值处理的三个法宝：`isnull()` 找缺失、`fillna()` 填缺失、`drop()` 删列。


### 3.2 练习：数据清洗 — 缺失值处理 <span class="badge-core">必做</span>

1. 检查每列有多少缺失值
2. 将 `PassengerId` 设为行索引
3. `Age` 列缺失值用 **中位数** 填充
4. `Embarked` 列缺失值用 **众数** (mode) 填充
5. `Cabin` 列缺失过多（>70%），直接删除该列
6. 再次确认无缺失值

````python
# 请在此处编写代码
import pandas as pd
pd.set_option('display.width', 200)
pd.set_option('display.max_columns', None)
titanic=pd.read_csv('titanic.csv')
print('Titanic(before data cleaning):\n',titanic)
l=titanic.isnull().sum()
print('数据清洗前各列缺失值的数量为:\n',l)

titanic.set_index('PassengerId',inplace=True)

titanic['Age']=titanic['Age'].fillna(titanic['Age'].median())

titanic['Embarked']=titanic['Embarked'].fillna(titanic['Embarked'].mode()[0])

titanic.drop('Cabin',axis=1,inplace=True)

print('Titanic(after data cleaning):\n',titanic)
L=titanic.isnull().sum()
print('数据清洗后各列缺失值的数量为:\n',L)
if L.any():
    print('有缺失值')
else:
    print('无缺失值')
````

<div class="hint-box">
<details>
<summary>💡 点击查看提示</summary>

- `.isnull().sum()` 统计每列缺失值个数
- `.set_index('列名', inplace=True)` 设置索引
- `.median()` 求中位数，`.mode()[0]` 求众数
- `.fillna(值, inplace=True)` 填充缺失值
- `.drop('列名', axis=1, inplace=True)` 删除列
</details>

## 4. 条件筛选与统计分析

### 4.1 知识准备：筛选与排序

````python
# 创建一个示例 DataFrame
df = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana'],
    'Score': [85, 92, 78, 95],
    'Grade': ['B', 'A', 'C', 'A']
})
df
````

````python
# ===== 布尔索引：筛选符合条件的行 =====
df[df['Score'] > 80]
````

````python
# ===== 排序 =====
df.sort_values('Score', ascending=False)
````

````python
# ===== apply()：对每行/每列应用自定义函数 =====
def grade_to_gpa(grade):
    mapping = {'A': 4.0, 'B': 3.0, 'C': 2.0}
    return mapping.get(grade, 0)

df['GPA'] = df['Grade'].apply(grade_to_gpa)
df
````

> ✅ 三个核心操作：`df[条件]` 筛选行、`.sort_values()` 排序、`.apply()` 逐元素处理。


### 4.2 练习：条件筛选与统计分析 <span class="badge-core">必做</span>

1. 有多少人生还？生还率是多少？
2. 找出年龄最大和年龄最小的乘客，打印他们的姓名和年龄
3. 按船票价格 **降序** 排列，显示最贵的 **10** 张票的乘客姓名与票价
4. 编写函数 `age_group(age)`：<18 返回 `"Child"`，18–60 返回 `"Adult"`，>60 返回 `"Senior"`。用 `.apply()` 新建一列 `AgeGroup`，并统计各组人数

````python
# 请在此处编写代码
Sur_sum=titanic['Survived'].sum()
Sur_rate=Sur_sum/titanic.shape[0]
print('共有',Sur_sum,'人生还；','生还率为：',Sur_rate*100,'%')

Oldest=titanic['Age'].idxmax()
Youngest=titanic['Age'].idxmin()
print('年龄最大的乘客的姓名和年龄：',titanic.at[Oldest,'Name'],titanic.at[Oldest,'Age'],'\n年龄最小的乘客的姓名和年龄：',titanic.at[Youngest,'Name'],titanic.at[Youngest,'Age'])

titanic.sort_values('Ticket',ascending=False,inplace=True)
print('Titanic上船票价格前十的乘客姓名与票价为：\n',titanic.head(10)[['Name','Ticket']])

def age_group(age):
    Age_dic={0:'Child',1:'Adult',2:'Senior'}
    if (age<18):
        i=0
    elif (age<=60):
        i=1
    else:
        i=2
    return Age_dic.get(i,0)

titanic['AgeGroup']=titanic['Age'].apply(age_group)
print('新增AgeGroup统计项后的titanic:\n',titanic)

Age_Group={}
Age_Group['Child']=len(titanic[titanic['AgeGroup']=='Child'])
Age_Group['Adult']=len(titanic[titanic['AgeGroup']=='Adult'])
Age_Group['Senior']=len(titanic[titanic['AgeGroup']=='Senior'])
print('各年龄段统计结果:\n',Age_Group)
````

## 5. 分组聚合

### 5.1 知识准备：分组聚合

````python
# 创建一个产品销售数据
sales = pd.DataFrame({
    'Region': ['East', 'West', 'East', 'West', 'East', 'West'],
    'Product': ['A', 'A', 'B', 'B', 'A', 'B'],
    'Revenue': [100, 150, 200, 120, 180, 90],
    'Units': [10, 15, 20, 12, 18, 9]
})
sales
````

````python
# ===== 单列分组，单指标聚合 =====
sales.groupby('Region')['Revenue'].mean()
````

````python
# ===== 单列分组，多指标聚合 =====
sales.groupby('Region').agg(
    Total_Revenue=('Revenue', 'sum'),
    Avg_Units=('Units', 'mean'),
    Count=('Revenue', 'count')
)
````

````python
# ===== 透视表：行+列两个维度 =====
pd.pivot_table(sales, values='Revenue', index='Region', columns='Product', aggfunc='sum')
````

> ✅ `groupby().agg()` 是分组统计的瑞士军刀，`pivot_table()` 用于二维交叉分析。


### 5.2 练习：分组聚合 — 谁更可能生还？ <span class="badge-core">必做</span>

1. 按 **性别** 分组，计算每组的：生还率、平均年龄、平均票价、人数
2. 按 **船舱等级 (Pclass)** 分组，计算每组的生还率
3. 创建 **透视表**：行= Pclass，列= Sex，值= Survived 均值
4. 根据以上结果，哪个群体生还率最高？哪个最低？用代码输出结论

````python
# 请在此处编写代码
titanic_sex=titanic.groupby('Sex').agg(
    Sur_rate=('Survived','mean'),
    Avg_age=('Age','mean'),
    Ave_Fare=('Fare','mean'),
    Sum_count=('Name','count')
)
print('按性别分组，各组的相关指标:\n',titanic_sex)

titanic_Pclass=titanic.groupby('Pclass')['Survived'].mean()
print('按船舱分组，各组的生还率:\n',titanic_Pclass)

titanic_pivot=pd.pivot_table(titanic,values='Survived',index='Pclass',columns='Sex',aggfunc='mean')
print('透视表:\n',titanic_pivot)

Sur_max=titanic_pivot.stack().idxmax()
Sur_min=titanic_pivot.stack().idxmin()
print('生还率最高的群体是:',Sur_max[1],'from',Sur_max[0],'Class')
print('生还率最低的群体是:',Sur_min[1],'from',Sur_min[0],'Class')
````

- `pd.pivot_table(df, values='V', index='行', columns='列', aggfunc='mean')`

## 6. NumPy 向量化计算

### 6.1 知识准备：NumPy 向量化

````python
import numpy as np

# ===== 向量化操作：无需循环！ =====
arr = np.array([1, 2, 3, 4, 5])
print("原数组:", arr)
print("+10:", arr + 10)        # 每个元素 +10
print("平方:", arr ** 2)       # 每个元素平方
print("求和:", np.sum(arr))    # 聚合函数
````

````python
# ===== 向量化 vs 循环 速度对比 =====
big = np.random.rand(1_000_000)
print("NumPy (向量化):", end=" ")
%timeit np.mean(big)

def loop_mean(a):
    total = 0.0
    for x in a:
        total += x
    return total / len(a)
print("For 循环:       ", end=" ")
%timeit loop_mean(big)
````

````python
# ===== 广播 (Broadcasting) =====
matrix = np.array([[1, 2, 3], [4, 5, 6]])
print("原矩阵:\n", matrix)
row_means = np.mean(matrix, axis=1, keepdims=True)
print("\n每行均值:\n", row_means)
print("\n减去行均值 (广播):\n", matrix - row_means)
````

> ✅ NumPy 三要素：**向量化操作**、**广播机制**、**丰富的数学函数**。


### 6.2 练习：NumPy 数组操作 <span class="badge-core">必做</span>

1. 提取 `Age` 列（去除NaN）为 NumPy 数组。分别用 **for循环** 和 **NumPy函数** 计算平均值，用 `%timeit` 比较耗时
2. 提取 `Age` 和 `Fare` 两列组成二维数组（去除含NaN的行）。打印数组的 **形状、均值、标准差**
3. 对该数组做 **Z-score 标准化**（每列减均值除以标准差），验证标准化后均值≈0、标准差≈1
4. 使用 `np.corrcoef()` 计算 Age 与 Fare 的 **相关系数矩阵***

````python
# 请在此处编写代码
Ages=titanic['Age'].to_numpy()
print('Numpy计算的均值:',np.mean(Ages),'耗时为:')
%timeit np.mean(Ages)

def agemean(array):
    sum=0
    for i in array:
        sum+=i
    return sum/(len(array))
print('for计算的均值:',agemean(Ages),'耗时为:')
%timeit agemean(Ages)

AnF=titanic[['Age','Fare']].to_numpy()
AnF_avg=np.nanmean(AnF,axis=0)
AnF_std=np.nanstd(AnF,axis=0)
print('shape:',AnF.shape)
print('Age_avg Fare_avg\n',AnF_avg)
print('Age_std Fare_std\n',AnF_std)
Zs_AnF=(AnF-AnF_avg)/AnF_std
print('the mean of AnF(after Z-score):\n',np.nanmean(Zs_AnF,axis=0))
print('the std of AnF(after Z-score):\n',np.nanstd(Zs_AnF,axis=0))
corr=np.corrcoef(AnF,rowvar=False)
print('Age和Fare的相关系数矩阵为:\n',corr)
````

<div class="hint-box">
<details>
<summary>💡 点击查看提示</summary>

- `.dropna().to_numpy()` 将Pandas列转为NumPy数组
- `%timeit` 是Jupyter魔法命令，用于测量代码执行时间
- `np.mean(arr, axis=0)` 沿行方向（对每列）求均值
- `np.std(arr, axis=0)` 同理求标准差
- `np.corrcoef(arr.T)` 计算列之间的相关系数（需转置）
- 广播机制：`(arr - mean) / std` 自动对每列操作
</details>

使用以下参数方程在 $t \in [0, 2\pi]$ 上绘制曲线：

$$\begin{cases} x = 16 \sin^3(t) \\ y = 13\cos(t) - 5\cos(2t) - 2\cos(3t) - \cos(4t) \end{cases}$$

- 用 `np.linspace` 生成 1000 个均匀分布的 $t$ 值
- 用红色线条绘制，标题为 "Heart Curve ♥"
- 设置 `plt.axis('equal')` 保持比例

````python
# 请在此处编写代码
t=np.linspace(0,2*np.pi,1000)

x=16*np.sin(t)
y=13*np.cos(t)-5*np.cos(2*t)-2*np.cos(3*t)-np.cos(4*t)

plt.figure(figsize=(6,6))
plt.fill(x,y,color='red',linewidth=2,alpha=0.5)
plt.title('Heart Curve ♥')
plt.axis('equal')
plt.grid()
plt.show()
````
